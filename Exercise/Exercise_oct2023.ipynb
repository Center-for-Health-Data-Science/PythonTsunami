We will investigate what data is missing and try to impute it.

A word of caution:

Note that imputation is a __complex subject__ and whether it makes sense to do it and the method used highly depend on the data set. Sometimes, the mean of a value across all non-missing observations is a good approximation for the missing value. On the other hand, if you have a column that says whether or not the person was treated with the drug or the placebo we have no good way to guess which treatment the person received. Replacing missing values in this column with the most common value (which is that they did get the drug) will produce extremely __wrong data__ and lead you to __wrong conclusions__. Do not do that!

But, before we go on and interpret our results, we need to check the assumptions our ANOVA makes.

#testing normality
stats.shapiro(model.resid)

We need to reject the assumption of normality that the Shapiro-Wilk test tests for. However, we can also do a visual inspection of our data

````python
#visual check for normality 

fig = plt.figure(figsize= (10, 10))
ax = fig.add_subplot(111)

normality_plot, stat = stats.probplot(model.resid, plot= plt, rvalue= True)
ax.set_title("Probability plot of model residual's", fontsize= 20)
ax.set

plt.show()
````

In our probability plot we can see that our residuals correlate well with a theoretical normal distribution, but there is a bit of a curve visible.

````python
#check homogeneity of variance using Levene's test
stats.levene(df_impute['Albumin'][df_impute['Stage'] == 1.0],
               df_impute['Albumin'][df_impute['Stage'] == 2.0],
               df_impute['Albumin'][df_impute['Stage'] == 3.0],
               df_impute['Albumin'][df_impute['Stage'] == 4.0])
````

Levene's test is non-significant, indicatint that there is non-significant difference in the variance between the 4 stages, but we can check this visually as well.

````python
#visual check for homogeneity of variance
fig = plt.figure(figsize= (10, 10))
ax = fig.add_subplot(111)

ax.set_title("Box Plot of Albumin by Stage", fontsize= 20)
ax.set

data = [df_impute['Albumin'][df_impute['Stage'] == 1.0],
               df_impute['Albumin'][df_impute['Stage'] == 2.0],
               df_impute['Albumin'][df_impute['Stage'] == 3.0],
               df_impute['Albumin'][df_impute['Stage'] == 4.0]]

ax.boxplot(data,
           tick_labels= ['1', '2', '3', '4'],
           showmeans= True)

plt.xlabel("Stage")
plt.ylabel("Albumin")

plt.show()
````

The visual inspection of the homogeneity of variance looks good as well, so it is reasonable to our sample groups have equal variance. Even though our residuals seemed to mostly follow the theoretical normal distribution, as the Shapiro-Wilk test was significant we can decide to change our one-way ANOVA to a Kruskal-Wallis ANOVA. (But it would not have been unreasonable to keep the original one-way ANOVA.)

````python
stats.kruskal(df_impute['Albumin'][df_impute['Stage'] == 1.0],
               df_impute['Albumin'][df_impute['Stage'] == 2.0],
               df_impute['Albumin'][df_impute['Stage'] == 3.0],
               df_impute['Albumin'][df_impute['Stage'] == 4.0])
````
