**Question 2**

Consider the data on monkeypox cases in the EU/EEA in the following file: https://opendata.ecdc.europa.eu/monkeypox/casedistribution/csv/data.csv 
How do you:

1. Read the data into pandas?
2. Display all column names?
3. Sort the number of confirmed cases from highest to lowest?
4. Find which country had the maximum number of newly recorded cases on one day?
5. Create new columns for day, month, and year? 
>Hint, you can use str.split() for this.

6. Calculate for Portugal how many confirmed cases there were in total per month and the daily maximum per month?

```python
#1
url_ecdc_monkeypox = "https://opendata.ecdc.europa.eu/monkeypox/casedistribution/csv/data.csv"
df = pd.read_csv(url_ecdc_monkeypox)
```

```python
#2
df.columns
```

```python
#3
df.sort_values(by='ConfCases', ascending=False).head(10)
```

```python
#4
df.loc[df['ConfCases']==df['ConfCases'].max(), ['DateRep', 'CountryExp', 'ConfCases']]
```

```python
#5
df[['year', 'month', 'day']] = df['DateRep'].str.split('-', expand=True)
```

```python
#6
df.loc[df['CountryExp']=='Portugal'].groupby(['month', 'year']).agg({'ConfCases':['sum','max']})
```
