```python
# Import modules from 'matplotlib'.
import matplotlib.pyplot as plt

# Import modules from 'seaborn'.
import seaborn as sns
```

```python
# Example dataset: health expenditure & life expectancy for a few
# countries.
#
# Columns: Year, Country, Spending_USD, Life_Expectancy
healthexp = sns.load_dataset("healthexp").sort_values("Year")

# Inspect the first lines of the data frame.
healthexp.head()

# Second example dataset: tips.
tips = sns.load_dataset("tips")

# Inspect the first lines of the data frame.
tips.head()
```

### Exercise 1

```python
# Solution: line plot with two countries

countries = ["France", "Germany"]
df_two = healthexp[healthexp["Country"].isin(countries)]

fig, ax = plt.subplots(figsize=(6, 4))

sns.lineplot(
    data=df_two,
    x="Year",
    y="Spending_USD",
    hue="Country",
    marker="o",
    ax=ax,
)

ax.set_title("Health spending over time – two countries")
ax.set_xlabel("Year")
ax.set_ylabel("Health spending per person (USD)")

# Make sure the y-axis starts at 0
ax.set_ylim(bottom=0)

plt.show()
```

### Exercise 2

```python
tips["tip_pct"] = tips["tip"] / tips["total_bill"] * 100

fig, ax = plt.subplots(figsize=(6, 4))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip_pct",
    hue="smoker",
    ax=ax,
)

ax.set_title("Tip percentage vs total bill, colored by smoker")
ax.set_xlabel("Total bill")
ax.set_ylabel("Tip (%)")

plt.show()
```

### Exercise 3

```python
mean_tip_by_day = tips.groupby("day")["tip"].mean().sort_values()
order = mean_tip_by_day.index  # order of days from lowest to highest mean tip

fig, ax = plt.subplots(figsize=(6, 4))

sns.barplot(
    data=tips,
    x="day",
    y="tip",
    order=order,
    ax=ax,
)

ax.set_title("Average tip by day (sorted)")
ax.set_ylabel("Average tip ($)")
ax.tick_params(axis="x", rotation=45)

plt.show()
```

### Exercise 4

```python
latest_year = healthexp["Year"].max()
df_latest = healthexp[healthexp["Year"] == latest_year]

fig, ax = plt.subplots(figsize=(6, 4))

sns.histplot(
    data=df_latest,
    x="Life_Expectancy",
    bins=15,
    stat="density",
    ax=ax,
)

mean_le = df_latest["Life_Expectancy"].mean()
ax.axvline(mean_le, linestyle="--")

ax.set_title(f"Life expectancy in {latest_year}")
ax.set_xlabel("Life expectancy (years)")

plt.tight_layout()
plt.show()
```

### Exercise 5

```python
tips = tips.copy()
tips["tip_pct"] = tips["tip"] / tips["total_bill"] * 100

fig, ax = plt.subplots(figsize=(7, 4))

sns.violinplot(
    data=tips,
    x="day",
    y="tip_pct",
    hue="time",
    split=True,
    inner="quartile",
    ax=ax,
)

ax.set_title("Tip percentage by day and time")
ax.set_ylabel("Tip (%)")
ax.legend(title="Time of day")

plt.tight_layout()
plt.show()
```

```python
# Option B.

corr_tips = tips[["total_bill", "tip", "size"]].corr()

fig, ax = plt.subplots(figsize=(4, 4))

sns.heatmap(
    corr_tips,
    annot=True,
    fmt=".2f",
    cmap="RdBu_r",
    center=0,
    vmin=-1,
    vmax=1,
    ax=ax,
)

ax.set_title("Correlation: total bill, tip, size")

plt.show()
```
