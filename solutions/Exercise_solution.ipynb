### 2.1 Solution: Violin plots and histograms


Consider you dataframe columns that are not the outcome variable. How are the measurements distributed?

To study the distributions we want to make **violin plots** of variables, i.e. data columns, that are numeric and **histograms** of the variables that are strings/categorical.

To check for the data type of a column, have a look at how data types are specified in `dataframe.dtypes`. Then check the data type of each column. Remember a column is a **pandas `Series`**, so it has a `dtype` attribute (only dataframes have `dtypes`!).

You can start by figuring out how to make a plot of the data in one column. Once you have that, make one plot for each column that is numeric or a string (except the outcome).

**Pro version**: Some columns are not actually predictors, such as a the ID column. You can identify these columns i.e. by seeing that each of their values is unique (this would be very unlikely for a measured variable). Skip them when making the plots.


````python
import matplotlib.pyplot as plt
import seaborn as sns
from pandas.api.types import is_numeric_dtype

# set a basic seaborn style (optional)
sns.set_theme(style="whitegrid")

for col_name in data.columns:
    # skip ID column
    if col_name == "id":
        continue

    if is_numeric_dtype(data[col_name]):
        # NUMERIC: violin plot with box + all data points

        fig, ax = plt.subplots(figsize=(5, 4))

        # violin with a box in the middle
        sns.violinplot(
            y=data[col_name],
            inner="box",     # show a boxplot inside the violin
            ax=ax
        )

        # all points on top
        sns.stripplot(
            y=data[col_name],
            color="black",
            alpha=0.4,
            size=2,
            ax=ax
        )

        ax.set_title(col_name)
        ax.set_ylabel(col_name)

        plt.tight_layout()
        plt.show()

    else:
        # NON-NUMERIC: histogram / count plot, separated by category

        fig, ax = plt.subplots(figsize=(5, 4))

        # categorical “histogram”: bar for each category,
        # colored by the category itself
        sns.histplot(
            data=data,
            x=col_name,
            hue=col_name,     # separate by different values (colors)
            multiple="stack",
            discrete=True,
            ax=ax
        )

        ax.set_title(col_name)
        ax.set_xlabel(col_name)
        ax.legend(title=col_name)

        plt.tight_layout()
        plt.show()
````

We could also choose to color the plot by the outcome variable, stroke, so we get two plots for the two different outcomes.

````python
import matplotlib.pyplot as plt
import seaborn as sns
from pandas.api.types import is_numeric_dtype

sns.set_theme(style="whitegrid")

for col_name in data.columns:
    # skip the ID and target column
    if col_name in ("id", "stroke"):
        continue

    if is_numeric_dtype(data[col_name]):
        # NUMERIC: two violins, one for stroke == 0 and one for stroke == 1
        fig, ax = plt.subplots(figsize=(5, 4))

        sns.violinplot(
            data=data,
            x="stroke",       # separate violins by stroke
            y=col_name,
            ax=ax
        )

        ax.set_title(col_name)
        ax.set_xlabel("stroke")
        ax.set_ylabel(col_name)

        plt.tight_layout()
        plt.show()

    else:
        # NON-NUMERIC: histogram grouped by stroke, using proportions
        fig, ax = plt.subplots(figsize=(5, 4))

        sns.histplot(
            data=data,
            x=col_name,
            hue="stroke",
            multiple="dodge",      # side-by-side bars (like barmode='group')
            stat="probability",    # like histnorm='probability'
            common_norm=False,     # don’t force groups to sum to 1 together
            discrete=True,         # treat as categorical
            ax=ax
        )

        ax.set_title(col_name)
        ax.set_xlabel(col_name)
        ax.legend(title="stroke")

        plt.tight_layout()
        plt.show()
````

A word of caution:

Note that imputation is a __complex subject__ and whether it makes sense to do it and the method used highly depend on the data set. Sometimes, the mean of a value across all non-missing observations is a good approximation for the missing value. On the other hand, if you have a column that says whether or not the person was treated with the drug or the placebo we have no good way to guess which treatment the person received. Replacing missing values in this column with the most common value (which is that they did get the drug) will produce extremely __wrong data__ and lead you to __wrong conclusions__. Do not do that!

3. Create both a histogram and a Q–Q plot for Albumin levels and assess whether the data appear approximately normally distributed. 

    The histogram can be generated using Seaborn, as before. 
    In principle, a Q–Q plot could be constructed from scratch by computing empirical quantiles and visualizing them with a Seaborn scatterplot. However, this is unnecessary here, as we have already imported scipy.stats, which provides a built-in function for generating Q–Q plots.

    `stats.probplot()`

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Left: Histogram
sns.histplot(
    data["Albumin"].dropna(),
    kde=True,
    ax=axes[0]
)
axes[0].set_title("Histogram of Albumin")

# Right: Q–Q plot
stats.probplot(
    data["Albumin"].dropna(),
    plot=axes[1]
)
axes[1].set_title("Q–Q plot of Albumin")

plt.tight_layout()
plt.show()
````

To complement the visual assessment of the data distribution, we can perform a Shapiro–Wilk test. This test has the null hypothesis (H₀) that the data are drawn from a normally distributed population.

A p-value greater than 0.05 indicates that we cannot reject the null hypothesis (i.e. the data are consistent with normality), whereas a p-value less than 0.05 suggests that we reject H₀ and conclude that the data deviate from normality.

`stats.shapiro()`

**Note:** The Shapiro–Wilk test is sensitive to both sample size and the presence of skewness or heavy tails. As a result, conclusions about normality should not be based on this test alone, but interpreted in conjunction with visual diagnostics such as histograms and Q–Q plots.

4. Run the Shapiro-wilk test for Albumin. What do you conclude based on the plots you have made and the p-value from the test. Are the data normally distibuted?

````python
stats.shapiro(data["Albumin"].dropna())
````

### 4.1 One-way ANOVA (Analysis of Variance) 
Having examined the distribution of Albumin, we conclude that it is approximately normally distributed. We therefore proceed with a one-way ANOVA to determine whether the mean level of this protein differs significantly across one or more of the four Cirrhosis stages.

Before performing the analysis, we must first arrange the data in the format required by the one-way ANOVA function.

1. Make a list containing Albumin values per stage (list of length four). Ensure that you do not have any missing values in your dataset, i.e. filter these out.


    HINT: Loop over stages and extract Albumin values for each of these. Remember to drop missing values.

````python
albu_list = []

for stage in data["Stage"].unique().dropna():
    
    # select bilirubin values for this stage
    albu = data.loc[data["Stage"] == stage, "Albumin"].dropna()
    
    # store results
    albu_list.append(albu)

print(albu_list)
````

2. Use the list you generated above for One-Way ANOVA. See pseudo code below! What does your p-value tell your about any potential differernces between group means?

    `stats.f_oneway(*My_List)`

````python
an_model = stats.f_oneway(*albu_list)

print(an_model)
````

### 4.2 Logistic regression
In ANOVA, we test for differences in group means. While this can be useful, it is often too limited for real-world data, where we want to model outcomes in a continuous framework and account for covariates and potential confounders.

In the simplest case with two groups (e.g., control vs. treatment), a regression-based approach like logistic regression for a binary outcome or linear regression for a continuous outcome—provides a more flexible alternative. When the outcome has more than two categories (stages 1–4), this extends to multinomial regression, which allows us to model category membership while adjusting for additional predictors.

In the below we will perform logistic regression, using the patient suvival status as the outcome - N.B this is not survival analysis! 

1. Make a new dataset which contains only; `Status`, `Bilirubin`, `Age`, `Sex`. We only want complete data, so drop missing values.

````python
df_logistic = data.loc[:, ["Status", "Bilirubin", "Age", "Sex"]].dropna()
````

2. Make a new column named `Deceased`. Patients which have the label `D` in the `Status` column should be encoded with a **1 (D = 1)** in the new `Deceased` column, while anything else should be encoded as **0**. You need to ensure that the data type of this new column is integer!

````python
df_logistic["Deceased"] = (df_logistic["Status"] == "D").astype(int)

print(df_logistic)
````

3. Run the logistic regression. You should have `Deceased` as your outcome and `Bilirubin`, `Age` and `Sex` as your predictors. After making your model you call `.fit()` on it to estimate the coefficients. Any categorical predictor variables must be wrapped in `C()` to ensure they are treated as categorical variables by the model. See pseudo code:

 `model = logit('Outcome ~ V1 + V2 + ...', data=dataset)`

 `modelFit = model.fit()`

 

````python
logit_model = logit('Deceased ~ Bilirubin + Age + C(Sex)', data=df_logistic)

modelFit = logit_model.fit()
````
