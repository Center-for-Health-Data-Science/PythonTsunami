### Exercise 3

Write a function that returns **both** the sum of the first two inputs and the difference between the second and third input: 

```python
def sum_dff(x, y, z):
    sum_xy = x + y 
    diff_yz = abs(y - z)
    return sum_xy, diff_yz
```

```python
# define the input for your function
a = 10
b = 15
c = 20
```

```python
sum_dff(a,b,c)
```

### Exercise 4

Write a function that takes the distance from home to university as input and returns a mode of transportation. For distances below 1 km it should suggest you to walk, for more than 12 km it should suggest the metro or train, otherwise it should suggest you to take your bike. 

    
> Tip: use a descriptive function name (in lowercase with words separated by underscores if neccesary to improve readability)

```python
def transport(distance):
    if distance < 1:
        return 'walk'
    elif distance > 12:
        return 'metro or train'
    else:
        return 'bike'
```

```python
transport(0.5)
```
